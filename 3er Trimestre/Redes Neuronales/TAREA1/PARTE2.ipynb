```python
import tensorflow as tf
from tensorflow.keras import layers, models

# 1. Cargar datos (esta vez sí usamos las etiquetas 'y' para clasificar)
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.fashion_mnist.load_data()
x_train, x_test = x_train / 255.0, x_test / 255.0

# 2. CARGAR el encoder de la Etapa 1
# Usamos la ruta de Drive que ya verificamos que funciona
ruta_modelo = '/content/drive/MyDrive/Maestría en Ciencias de la Inteligencia Artificial/3er Trimestre/Redes Neuronales/TAREA1/encoder_fashion_mnist.keras'
encoder = tf.keras.models.load_model(ruta_modelo)

# 3. CONGELAR EL ENCODER
# Esto asegura que los pesos que aprendió el Autoencoder no cambien
encoder.trainable = False 

# 4. Construir el modelo final
model = models.Sequential([
    encoder,                             # El encoder congelado
    layers.Flatten(),                    # Aplanamos el vector latente
    layers.Dense(128, activation='relu'),# Capa nueva para aprender a clasificar
    layers.Dropout(0.2),                 # Buena práctica para evitar sobreajuste
    layers.Dense(10, activation='softmax') # 10 clases (camisas, botas, etc.)
])

model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

# 5. Entrenar (entrena mucho más rápido porque el encoder no se mueve)
model.fit(x_train, y_train, epochs=10, validation_data=(x_test, y_test))
```

Git te está "frenando" el paso a la rama master porque detectó que el archivo PARTE1.ipynb tiene cambios locales que aún no guardaste en un commit dentro de tu rama actual (etapa2).

Esto sucede porque, como vimos antes, los notebooks suelen actualizar sus metadatos internos apenas los abrís o corrés una celda. Git no quiere que pierdas esos cambios al saltar a master.
